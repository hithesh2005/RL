```python
import random

machines = ['A', 'B', 'C']
epsilon = 0.2
plays = 10

N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': [], 'B': [], 'C': []}
Q = {'A': 0, 'B': 0, 'C': 0}

total_reward = 0

for i in range(plays):

    r = random.random()

    # Highest Q machine
    best = max(Q, key=Q.get)

    # Modified epsilon-greedy
    if r < epsilon:
        # Exploration: exclude highest Q machine
        other = [m for m in machines if m != best]
        machine = random.choice(other)
        method = "Exploration"
    else:
        # Exploitation: select highest Q machine
        machine = best
        method = "Exploitation"

    print("\nPlay", i + 1)
    print("Random value:", round(r, 3))
    print("Method:", method)
    print("Selected Machine:", machine)

    reward = float(input("Enter reward: "))

    # Update N
    N[machine] += 1

    # Store reward
    R[machine].append(reward)

    # Update Q using incremental average
    Q[machine] = Q[machine] + \
                 (reward - Q[machine]) / N[machine]

    total_reward += reward

    print("N(", machine, ") =", N[machine])
    print("Q(", machine, ") =", round(Q[machine], 2))
    print("Total Reward =", total_reward)

    print("Current Q values:")
    for m in machines:
        print(m, "=", round(Q[m], 2))


print("\n========== FINAL RESULTS ==========")

for machine in machines:
    print("\nMachine", machine)
    print("N =", N[machine])
    print("Rewards =", R[machine])
    print("Average Q =", round(Q[machine], 2))

print("\n===================================")
print("Total Reward =", total_reward)
print("Average Total Reward =", round(total_reward / plays, 2))

best = max(Q, key=Q.get)

print("Best Machine =", best)
print("Best Q Value =", round(Q[best], 2))
```
